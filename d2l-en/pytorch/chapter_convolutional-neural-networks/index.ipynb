# **Convolutional Neural Networks**
:label:`chap_cnn`

- **Image data representation**:
  - Represented as a **2D grid of pixels**.
  - Each pixel corresponds to:
    - A **single numerical value** (monochrome).
    - Or **multiple values** (color).

- **Previous approach**:
  - We **flattened images** into one-dimensional vectors.
  - This ignored the **spatial structure between pixels**.
  - Flattening was necessary to feed data into a **fully connected MLP**.

- **Issue with flattening**:
  - MLPs are **invariant to feature order**.
  - Permuting the pixel order yields **similar results**, which is:
    - **Unrealistic**, since **nearby pixels are related**.
  - Ideally, we want to **leverage spatial relationships** in image data.

- **Solution: Convolutional Neural Networks (CNNs)**:
  - Introduced in :cite:`LeCun.Jackel.Bottou.ea.1995`.
  - Designed to **capture spatial structure** in images.
  - **Ubiquitous** in computer vision tasks.

- **Impact of CNNs**:
  - Improved performance on datasets like **ImageNet** :cite:`Deng.Dong.Socher.ea.2009`.
  - **Convnets** led to **significant breakthroughs** :cite:`Krizhevsky.Sutskever.Hinton.2012`.

- **Design inspirations**:
  - Rooted in **biology**, **group theory**, and **experimentation**.

- **Advantages of CNNs**:
  - **Sample-efficient**.
  - **Computationally efficient**:
    - Fewer parameters than MLPs.
    - Easy to **parallelize on GPUs** :cite:`Chetlur.Woolley.Vandermersch.ea.2014`.

- **Beyond image data**:
  - Applied to:
    - **Audio** :cite:`Abdel-Hamid.Mohamed.Jiang.ea.2014`.
    - **Text** :cite:`Kalchbrenner.Grefenstette.Blunsom.2014`.
    - **Time series** :cite:`LeCun.Bengio.ea.1995`.
    - **Graph data** :cite:`Kipf.Welling.2016`.
    - **Recommender systems**.
