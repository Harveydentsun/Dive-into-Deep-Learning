# **Builders' Guide**
:label:`chap_computation`

- **Role of Software Tools in Deep Learning**:
  - Alongside **giant datasets** and **powerful hardware**, great **software tools** have been essential for deep learning progress.
  - Early Example: Theano library (released in 2007) enabled researchers to **prototype models rapidly**.

- **Evolution of Deep Learning Libraries**:
  - From **low-level implementations** to **high-level abstractions**.
  - Transition from specifying **individual neurons** to designing **whole layers and blocks**.

- **Previous Chapters Recap**:
  - Introduced **basic machine learning concepts** and built **fully-functional deep learning models**.
  - Implemented **MLPs from scratch** and using **high-level APIs**.

- **This Chapter Focuses On**:
  - Exploring **key components of deep learning computation**, including:
    - **Model construction**.
    - **Parameter access and initialization**.
    - **Designing custom layers and blocks**.
    - **Reading and writing models to disk**.
    - **Leveraging GPUs** for speed improvements.
  - Transitioning from **end user** to **power user**:
    - Utilizing **mature deep learning libraries** while retaining flexibility for **custom implementations**.

- **Future Relevance**:
  - Techniques covered here are foundational for **advanced modeling chapters** that follow.
